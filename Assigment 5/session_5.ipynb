```python
# Let's create a CSV (comma separated values) file
header = "Name,Age,Grade\n"
rows = [
    "Jaume,30,8.9\n",
    "Francisco,25,7.1\n",
    "Elena,35,9.2\n"
]
```

```python
with open("grades.csv", "w") as f:
    f.write(header) # Write the header
    
    # Write each row of data
    for row in rows:
        f.write(row)
```

```python
with open("grades.csv", "r") as f:
    lines = f.read().splitlines()
    
header = lines.pop(0)
header = header.split(',')

print(header)

grades = {'students': []}
# create dictionary
for line in lines:
    student_dict = {}
    values = line.split(',')
    for idx, column in enumerate(header):
        student_dict[column] = values[idx]
    grades['students'].append(student_dict)
    
grades
```

## Basic Libraries II

Similar to JSON, Python includes a Pickle library. However, in contrast to the JSON format, Pickle is a Python-specific serialization format. The Pickle library provides tools to serialize Python objects, which involves transforming them into a stream of bytes. It also allows you to read these byte streams by deserializing them, transforming them back into their original Python objects.

In contrast to the JSON format, the binary format is usually more compact and, therefore, more efficient.

```python
import numpy as np
data = np.random.rand(10)

import pickle

# Serializing (dumping) the object
with open('data.pkl', 'wb') as f:
    pickle.dump(data, f)

# Deserializing (loading) the object
with open('data.pkl', 'rb') as f:
    loaded_data = pickle.load(f)

print(loaded_data)
```

```python
import re
import glob
import os

# Regex pattern, r in front of strings tell python to treat them as raw strings
# we do this so slashes don't get interpret as scaping symbol
pattern = r'(\d{8})_(\d{6})_SN(\d+)_QUICKVIEW_VISUAL_([\d_]+)_([A-Za-z0-9\-_.]+)\.txt' 

annotations = glob.glob('session_4/annotations/*.txt')

for annotation in annotations:

    # extract the file name
    filename = os.path.basename(annotation)
    
    # Search and extract values
    match = re.match(pattern, filename)
    if match:
        date, time, satellite_number, version, unique_region = match.groups()
        print(f"Date: {date}; Time: {time}; SN: {satellite_number}; ver: {version}; region: {unique_region}")
```

Let's now try to use them to order the annotations by date

```python
import re
import glob
import os
from datetime import datetime

# Regex pattern, r in front of strings tell python to treat them as raw strings
# we do this so slashes don't get interpret as scaping symbol
pattern = r'(\d{8})_(\d{6})_SN(\d+)_QUICKVIEW_VISUAL_([\d_]+)_([A-Za-z0-9\-_.]+)\.txt' 

annotations = glob.glob('session_4/annotations/*.txt')

# let's create a dictionary where per each annotations we gather the datetime object
ann_datetime = []

for annotation in annotations:

    # extract the file name
    filename = os.path.basename(annotation)
    
    # Search and extract values
    match = re.match(pattern, filename)
    if match:
        date, time, _, _, _ = match.groups()

        # Put them together, e.g. "20240101192856"
        datetime_str = date + time 

        # Parse the string into a datetime object
        datetime_obj = datetime.strptime(datetime_str, "%Y%m%d%H%M%S")

        # Output the datetime object
        print(f"Datetime Object: {datetime_obj}")
        
        ann_datetime.append((filename, datetime_obj))
```

```python
indices = np.argsort([date for name, date in ann_datetime])
indices
```

```python
for i in indices:
    print(ann_datetime[i][0])
```

```python
#Here we import the necessary libraries
import os
from collections import defaultdict

#Here we paste the folder where the annotations are stored (URL) in my computer
annotations_folder = r"C:\Users\enric\OneDrive - Universitat Ramón Llull\ESADE\Term 1\Python for Data Science\Assigment 4\extracted_session_4\session_4\annotations"

#Here we aim to count the number of files per month-year
annotations_per_month = defaultdict(int)

# Pas 1: we aim to list all files in the folder
try:
    files = os.listdir(annotations_folder)
    print(f"We found a total of {len(files)} files in the annotations folder.")
    print()
except FileNotFoundError:
    print("The folder does not exist.")
    files = []

# Pas 2: Here we loop through each file and check if it follows the naming convention
for file in files:
    # Here we check if the filename starts with a date in YYYYMMDD format 
    if len(file) > 8 and file[:8].isdigit():
        date_part = file[:8]  # Extract date (Year and Month)
        year = date_part[:4]  # First 4 characters are the year
        month = date_part[4:6]  # Next 2 characters are the month
        
        # Here we increment the count for this month-year in the dictionary
        annotations_per_month[f"{year}-{month}"] += 1

# Pas 3: We identify the month with the highest count of annotations
if annotations_per_month:
    # Here we find the month with the maximum number of annotations
    most_annotations_month = max(annotations_per_month, key=annotations_per_month.get)
    most_annotations_count = annotations_per_month[most_annotations_month]
    print(f"The month with the most annotations is: {most_annotations_month} ({most_annotations_count} files)")
    print()
else:
    print("No valid annotation files found.")

# Finally we print the output:
print("The annotations per month are the following:")
for month, count in annotations_per_month.items():
    print(f"{month}: {count}")
```

```python
#Exercise 2a)

#Here we import the necessary libraries
import os
from collections import Counter
from datetime import datetime

#Here we initialize the counter of anotations per month
annotations_per_month = Counter()

# Here similar to the previous exercise we Loop through each file in the folder
for filename in os.listdir(annotations_folder):
    if filename.endswith(".txt"):  # We make sure that it's an annotation file
        # Here we extract date part from the filename 
        date_str = filename.split('_')[0]
        # Parse date into datetime object
        date = datetime.strptime(date_str, "%Y%m%d")
        # Use (year, month) as key for counting
        annotations_per_month[(date.year, date.month)] += 1

# Output: 
most_common_month = annotations_per_month.most_common(1)[0]
print("The annotations per month and year are the following:", annotations_per_month)
print("The Month with most annotations is:", most_common_month)
```

```python
#Exercise 2b) I did the Jason and Pickle version in this exercice

#Here we import the necessary libraries
import json
import pickle

# First we initialize the dictionary annotations_by_month
annotations_by_month = {}

# Here we populate the dictionary with file names grouped by year and month
for filename in os.listdir(annotations_folder):
    if filename.endswith(".txt"):
        # Here we extract date part from filename
        date_str = filename.split('_')[0]
        date = datetime.strptime(date_str, "%Y%m%d")
        # Here we convert the year, month tuple into a string "YYYY-MM"
        key = f"{date.year}-{date.month:02d}"
        
        # Here we need to append filename to the appropriate month key in annotations_by_month
        if key not in annotations_by_month:
            annotations_by_month[key] = []
        annotations_by_month[key].append(filename)

# Now we need to save  the dictionary in JSON format
with open("annotations_by_month.json", "w") as json_file:
    json.dump(annotations_by_month, json_file)

# Here we load the JSON to verify that everything is ok
with open("annotations_by_month.json", "r") as json_file:
    loaded_json = json.load(json_file)
    print("Loaded JSON:", loaded_json)

# Now we do the same but using Pickle. 
with open("annotations_by_month.pkl", "wb") as pickle_file:
    pickle.dump(annotations_by_month, pickle_file)

# Here we do the same, load the Pickle to verify that everything is ok
with open("annotations_by_month.pkl", "rb") as pickle_file:
    loaded_pickle = pickle.load(pickle_file)
    print("Loaded Pickle:", loaded_pickle)


```

```python
# Exercise 3

# Here we initialize a list to store filtered annotations (only for the second half of 2024)
second_half_2024_annotations = []

# Now we loop through each annotation file in the annotations folder
for filename in os.listdir(annotations_folder):
    # Here we ensure it's a text file 
    if filename.endswith(".txt"):
        # Here we extract the date part of the filename, which is the first 8 characters 
        date_str = filename.split('_')[0]
        # Now we convert the date string into a datetime object
        date = datetime.strptime(date_str, "%Y%m%d")
        
        # Here we check if the date is in the second half of 2024 (July to December)
        if date.year == 2024 and date.month >= 7:
            # Here we append the filename and date as a tuple to the list for sorting
            second_half_2024_annotations.append((filename, date))

# Here we sort the annotations list by date from oldest to newest
second_half_2024_annotations.sort(key=lambda x: x[1])

# Now we print the sorted list of annotations or indicate if none are found.
if len(second_half_2024_annotations) > 0:
    print("Annotations from the second half of 2024, from oldest to newest:")
    for filename, date in second_half_2024_annotations:
        print(f"Annotation: {filename}, Date: {date}")
else:
    print("No annotations found in the second half of 2024.")
```
